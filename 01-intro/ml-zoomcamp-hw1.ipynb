```python
def vector_vector_multiplication(u, v):
    assert u.shape[0] == v.shape[0]

    result = 0.0

    for i in range(u.shape[0]):
        result += u[i] * v[i]

    return result
```

```python
a = np.full(3,2)
b = np.array([10,15,-2.3])
vector_vector_multiplication(a,b)
```
